```python
import pandas as pd

def analyze_ue(log_data):

    # =====================================================
    # STEP 1: Split input log into individual lines
    # =====================================================

    lines = log_data.strip().split("\n")

    records = []

    # =====================================================
    # STEP 2: Parse every log line
    # =====================================================

    for line in lines:

        parsed = parse_log_line(line)

        if parsed is not None:
            records.append(parsed)

    # If parser could not find valid records
    if len(records) == 0:

        return {
            "status": "error",
            "message": "No valid log records found"
        }

    # =====================================================
    # STEP 3: Create temporary DataFrame
    # =====================================================

    temp_df = pd.DataFrame(records)

    # Convert numeric columns
    temp_df["ue_id"] = pd.to_numeric(
        temp_df["ue_id"],
        errors="coerce"
    )

    temp_df["count"] = pd.to_numeric(
        temp_df["count"],
        errors="coerce"
    )

    # =====================================================
    # STEP 4: Failure detection
    # =====================================================

    failure_keywords = [
        "failure",
        "reject",
        "timeout",
        "error"
    ]

    failure_pattern = "|".join(failure_keywords)

    temp_df["is_failure"] = (
        temp_df["event"]
        .str.contains(
            failure_pattern,
            case=False,
            na=False
        )
    )

    # =====================================================
    # STEP 5: NLP EVENT CLASSIFICATION
    # =====================================================

    event_tfidf = loaded_vectorizer.transform(
        temp_df["event"]
    )

    temp_df["nlp_category"] = (
        loaded_nlp_model.predict(event_tfidf)
    )

    # =====================================================
    # STEP 6: UE LEVEL ANALYSIS
    # =====================================================

    results = []

    for ue_id, ue_df in temp_df.groupby("ue_id"):

        # -----------------------------
        # Basic features
        # -----------------------------

        total_events = len(ue_df)

        total_failures = int(
            ue_df["is_failure"].sum()
        )

        if total_events > 0:
            failure_rate = total_failures / total_events
        else:
            failure_rate = 0

        # -----------------------------
        # Authentication failures
        # -----------------------------

        authentication_failures = int(
            ue_df["event"]
            .str.contains(
                "Authentication Failure",
                case=False,
                na=False
            )
            .sum()
        )

        # -----------------------------
        # Handover failures
        # -----------------------------

        handover_failures = int(
            ue_df["event"]
            .str.contains(
                "Handover Failure",
                case=False,
                na=False
            )
            .sum()
        )

        # -----------------------------
        # Retransmission features
        # -----------------------------

        retransmission_df = ue_df[
            ue_df["event"].str.contains(
                "Retransmission",
                case=False,
                na=False
            )
        ]

        total_retransmissions = (
            retransmission_df["count"]
            .fillna(0)
            .sum()
        )

        avg_retransmission = (
            retransmission_df["count"]
            .mean()
        )

        if pd.isna(avg_retransmission):
            avg_retransmission = 0

        # =================================================
        # STEP 7: NLP category counts
        # =================================================

        nlp_counts = (
            ue_df["nlp_category"]
            .value_counts()
        )

        nlp_authentication_count = int(
            nlp_counts.get("Authentication", 0)
        )

        nlp_connection_count = int(
            nlp_counts.get("Connection", 0)
        )

        nlp_mobility_count = int(
            nlp_counts.get("Mobility", 0)
        )

        nlp_registration_count = int(
            nlp_counts.get("Registration", 0)
        )

        nlp_retransmission_count = int(
            nlp_counts.get("Retransmission", 0)
        )

        # =================================================
        # STEP 8: Create ML input
        # =================================================

        model_input = pd.DataFrame([{

            "total_events":
                total_events,

            "total_failures":
                total_failures,

            "avg_retransmission":
                avg_retransmission,

            "failure_rate":
                failure_rate,

            "authentication_failures":
                authentication_failures,

            "handover_failures":
                handover_failures,

            "total_retransmissions":
                total_retransmissions,

            "nlp_authentication_count":
                nlp_authentication_count,

            "nlp_connection_count":
                nlp_connection_count,

            "nlp_mobility_count":
                nlp_mobility_count,

            "nlp_registration_count":
                nlp_registration_count,

            "nlp_retransmission_count":
                nlp_retransmission_count

        }])

        # Keep EXACT feature order used during training
        model_input = model_input[
            loaded_metadata["feature_columns"]
        ]

        # =================================================
        # STEP 9: Root cause prediction
        # =================================================

        predicted_root_cause = (
            loaded_root_cause_model
            .predict(model_input)[0]
        )

        # =================================================
        # STEP 10: Final UE result
        # =================================================

        ue_result = {

            "ue_id":
                int(ue_id),

            "total_events":
                int(total_events),

            "total_failures":
                int(total_failures),

            "failure_rate":
                round(float(failure_rate), 4),

            "authentication_failures":
                authentication_failures,

            "handover_failures":
                handover_failures,

            "total_retransmissions":
                int(total_retransmissions),

            "avg_retransmission":
                round(float(avg_retransmission), 2),

            "nlp_authentication_count":
                nlp_authentication_count,

            "nlp_connection_count":
                nlp_connection_count,

            "nlp_mobility_count":
                nlp_mobility_count,

            "nlp_registration_count":
                nlp_registration_count,

            "nlp_retransmission_count":
                nlp_retransmission_count,

            "predicted_root_cause":
                str(predicted_root_cause)
        }

        results.append(ue_result)

    # VERY IMPORTANT
    return results
```

```python
def parse_log_line(line):

    line = line.strip()

    # Ignore empty lines
    if not line:
        return None

    # Split log using |
    parts = line.split("|")

    # Minimum required fields:
    # timestamp | RAT | protocol | event | UE
    if len(parts) < 5:
        return None

    timestamp = parts[0].strip()
    rat = parts[1].strip()
    protocol = parts[2].strip()
    event = parts[3].strip()

    # Extract UE ID
    ue_field = parts[4].strip()

    if "=" not in ue_field:
        return None

    ue_id = ue_field.split("=", 1)[1]

    # Optional fields
    cause = None
    count = None

    for field in parts[5:]:

        field = field.strip()

        if "=" not in field:
            continue

        key, value = field.split("=", 1)

        key = key.strip()
        value = value.strip()

        if key.lower() == "cause":
            cause = value

        elif key.lower() == "count":
            count = value

    return {
        "timestamp": timestamp,
        "rat": rat,
        "protocol": protocol,
        "event": event,
        "ue_id": ue_id,
        "cause": cause,
        "count": count
    }
```

```python
def generate_technical_summary(analysis):

    ue_id = analysis["ue_id"]
    total_events = analysis["total_events"]
    total_failures = analysis["total_failures"]
    failure_rate = analysis["failure_rate"]

    auth_failures = analysis["authentication_failures"]
    handover_failures = analysis["handover_failures"]

    total_retx = analysis["total_retransmissions"]

    root_cause = analysis["predicted_root_cause"]

    failure_percentage = failure_rate * 100

    summary = (
        f"UE {ue_id} contains {total_events} analyzed events, "
        f"with {total_failures} failure-related events "
        f"({failure_percentage:.1f}% failure rate). "
        f"The log contains {auth_failures} authentication failures, "
        f"{handover_failures} handover failures, and "
        f"{total_retx} total retransmissions. "
        f"The ML root-cause classifier identified "
        f"'{root_cause}' as the primary predicted category."
    )

    return summary
```

```python
def generate_recommendation(analysis):

    root_cause = analysis["predicted_root_cause"]

    if root_cause == "Authentication Failure":

        return (
            "Review authentication-related NAS signaling, "
            "associated cause codes, subscriber authentication "
            "parameters, and repeated authentication attempts."
        )

    elif root_cause == "Handover Failure":

        return (
            "Review mobility signaling, handover preparation and "
            "execution events, target-cell information, and "
            "handover-related cause codes."
        )

    elif root_cause == "High Retransmission":

        return (
            "Review MAC/RLC retransmission behavior and investigate "
            "radio conditions, link quality, and repeated transmission events."
        )

    elif root_cause == "High Failure Rate":

        return (
            "Investigate the dominant failure events and their associated "
            "protocol layers and cause codes."
        )

    else:

        return (
            "No major failure category was identified by the current model."
        )
```

```python
import pandas as pd

def analyze_ue(log_data):

    # -----------------------------------
    # 1. Parse raw log lines
    # -----------------------------------
    records = []

    for line in log_data.splitlines():

        result = parse_log_line(line)

        if result is not None:
            records.append(result)

    if len(records) == 0:
        return []

    temp_df = pd.DataFrame(records)

    # -----------------------------------
    # 2. Data type conversion
    # -----------------------------------
    temp_df["ue_id"] = pd.to_numeric(
        temp_df["ue_id"],
        errors="coerce"
    )

    temp_df["count"] = pd.to_numeric(
        temp_df["count"],
        errors="coerce"
    )

    # -----------------------------------
    # 3. Failure detection
    # -----------------------------------
    failure_keywords = [
        "failure",
        "reject",
        "timeout",
        "error"
    ]

    pattern = "|".join(failure_keywords)

    temp_df["is_failure"] = (
        temp_df["event"]
        .str.contains(
            pattern,
            case=False,
            na=False
        )
    )

    # -----------------------------------
    # 4. NLP Event Classification
    # -----------------------------------
    text_features = loaded_vectorizer.transform(
        temp_df["event"]
    )

    temp_df["nlp_predicted_category"] = (
        loaded_nlp_model.predict(
            text_features
        )
    )

    # -----------------------------------
    # 5. Create UE-level features
    # -----------------------------------
    results = []

    for ue_id, ue_df in temp_df.groupby("ue_id"):

        total_events = len(ue_df)

        total_failures = int(
            ue_df["is_failure"].sum()
        )

        failure_rate = (
            total_failures / total_events
            if total_events > 0
            else 0
        )

        authentication_failures = int(
            ue_df["event"]
            .str.contains(
                "Authentication Failure",
                case=False,
                na=False
            )
            .sum()
        )

        handover_failures = int(
            ue_df["event"]
            .str.contains(
                "Handover Failure",
                case=False,
                na=False
            )
            .sum()
        )

        total_retransmissions = (
            ue_df["count"]
            .fillna(0)
            .sum()
        )

        retransmission_rows = ue_df[
            ue_df["event"].str.contains(
                "Retransmission",
                case=False,
                na=False
            )
        ]

        avg_retransmission = (
            retransmission_rows["count"].mean()
            if len(retransmission_rows) > 0
            else 0
        )

        # -----------------------------------
        # NLP category counts
        # -----------------------------------
        category_counts = (
            ue_df["nlp_predicted_category"]
            .value_counts()
        )

        nlp_authentication_count = int(
            category_counts.get(
                "Authentication", 0
            )
        )

        nlp_connection_count = int(
            category_counts.get(
                "Connection", 0
            )
        )

        nlp_mobility_count = int(
            category_counts.get(
                "Mobility", 0
            )
        )

        nlp_registration_count = int(
            category_counts.get(
                "Registration", 0
            )
        )

        nlp_retransmission_count = int(
            category_counts.get(
                "Retransmission", 0
            )
        )

        # -----------------------------------
        # 6. Model input
        # -----------------------------------
        feature_values = {
            "total_events": total_events,
            "total_failures": total_failures,
            "avg_retransmission": avg_retransmission,
            "failure_rate": failure_rate,
            "authentication_failures":
                authentication_failures,
            "handover_failures":
                handover_failures,
            "total_retransmissions":
                total_retransmissions,
            "nlp_authentication_count":
                nlp_authentication_count,
            "nlp_connection_count":
                nlp_connection_count,
            "nlp_mobility_count":
                nlp_mobility_count,
            "nlp_registration_count":
                nlp_registration_count,
            "nlp_retransmission_count":
                nlp_retransmission_count
        }

        model_input = pd.DataFrame(
            [feature_values]
        )

        # Keep exact training feature order
        model_input = model_input[
            loaded_metadata["feature_columns"]
        ]

        # -----------------------------------
        # 7. Root cause prediction
        # -----------------------------------
        predicted_root_cause = (
            loaded_root_cause_model.predict(
                model_input
            )[0]
        )

        # -----------------------------------
        # 8. Final result
        # -----------------------------------
        results.append({
            "ue_id": int(ue_id),
            "total_events": total_events,
            "total_failures": total_failures,
            "failure_rate": float(failure_rate),
            "authentication_failures":
                authentication_failures,
            "handover_failures":
                handover_failures,
            "total_retransmissions":
                int(total_retransmissions),
            "avg_retransmission":
                float(avg_retransmission),
            "nlp_authentication_count":
                nlp_authentication_count,
            "nlp_connection_count":
                nlp_connection_count,
            "nlp_mobility_count":
                nlp_mobility_count,
            "nlp_registration_count":
                nlp_registration_count,
            "nlp_retransmission_count":
                nlp_retransmission_count,
            "predicted_root_cause":
                predicted_root_cause
        })

    return results


print("✅ analyze_ue() restored")
```

```python
parser_code = '''
def parse_log_line(line):
    """
    Parse one synthetic QXDM-style modem log line.

    Returns:
        dict: Structured log record
        None: If the line is invalid
    """

    line = line.strip()

    if not line:
        return None

    parts = line.split("|")

    if len(parts) < 5:
        return None

    timestamp = parts[0].strip()
    rat = parts[1].strip()
    protocol = parts[2].strip()
    event = parts[3].strip()

    ue_field = parts[4].strip()

    if "=" not in ue_field:
        return None

    ue_id = ue_field.split("=", 1)[1]

    cause = None
    count = None

    for field in parts[5:]:

        field = field.strip()

        if "=" not in field:
            continue

        key, value = field.split("=", 1)

        key = key.strip()
        value = value.strip()

        if key.lower() == "cause":
            cause = value

        elif key.lower() == "count":
            count = value

    return {
        "timestamp": timestamp,
        "rat": rat,
        "protocol": protocol,
        "event": event,
        "ue_id": ue_id,
        "cause": cause,
        "count": count
    }
'''

parser_path = (
    "/content/drive/MyDrive/"
    "QXDM_AI_Log_Analyzer/src/parser.py"
)

with open(parser_path, "w") as f:
    f.write(parser_code)

print("✅ parser.py created")
print(parser_path)
```

```python
inference_code = '''
import os
import json
import joblib
import pandas as pd

from parser import parse_log_line


PROJECT_PATH = "/content/drive/MyDrive/QXDM_AI_Log_Analyzer"
MODEL_PATH = os.path.join(PROJECT_PATH, "models")


# --------------------------------------------------
# Load saved ML/NLP artifacts
# --------------------------------------------------

vectorizer = joblib.load(
    os.path.join(MODEL_PATH, "tfidf_vectorizer.joblib")
)

nlp_model = joblib.load(
    os.path.join(MODEL_PATH, "nlp_classifier.joblib")
)

root_cause_model = joblib.load(
    os.path.join(MODEL_PATH, "root_cause_model.joblib")
)

with open(
    os.path.join(MODEL_PATH, "feature_metadata.json"),
    "r"
) as f:
    metadata = json.load(f)

feature_columns = metadata["feature_columns"]


# --------------------------------------------------
# UE analysis function
# --------------------------------------------------

def analyze_ue(log_data):

    records = []

    # Parse raw log text
    for line in log_data.strip().splitlines():

        parsed = parse_log_line(line)

        if parsed is not None:
            records.append(parsed)

    if not records:
        return []

    df = pd.DataFrame(records)

    # Convert numeric fields
    df["ue_id"] = pd.to_numeric(
        df["ue_id"],
        errors="coerce"
    )

    df["count"] = pd.to_numeric(
        df["count"],
        errors="coerce"
    )

    # Detect failure-related events
    failure_keywords = [
        "failure",
        "reject",
        "timeout",
        "error"
    ]

    pattern = "|".join(failure_keywords)

    df["is_failure"] = df["event"].str.contains(
        pattern,
        case=False,
        na=False
    )

    # --------------------------------------------------
    # NLP event classification
    # --------------------------------------------------

    text_features = vectorizer.transform(
        df["event"]
    )

    df["nlp_predicted_category"] = (
        nlp_model.predict(text_features)
    )

    results = []

    # --------------------------------------------------
    # Create features per UE
    # --------------------------------------------------

    for ue_id, ue_df in df.groupby("ue_id"):

        total_events = len(ue_df)

        total_failures = int(
            ue_df["is_failure"].sum()
        )

        failure_rate = (
            total_failures / total_events
            if total_events > 0
            else 0
        )

        authentication_failures = int(
            ue_df["event"]
            .str.contains(
                "Authentication Failure",
                case=False,
                na=False
            )
            .sum()
        )

        handover_failures = int(
            ue_df["event"]
            .str.contains(
                "Handover Failure",
                case=False,
                na=False
            )
            .sum()
        )

        retransmission_values = ue_df.loc[
            ue_df["event"].str.contains(
                "Retransmission",
                case=False,
                na=False
            ),
            "count"
        ].dropna()

        total_retransmissions = float(
            retransmission_values.sum()
        )

        avg_retransmission = (
            float(retransmission_values.mean())
            if len(retransmission_values) > 0
            else 0.0
        )

        # NLP category counts
        category_counts = (
            ue_df["nlp_predicted_category"]
            .value_counts()
        )

        feature_data = {
            "total_events": total_events,
            "total_failures": total_failures,
            "avg_retransmission": avg_retransmission,
            "failure_rate": failure_rate,
            "authentication_failures": authentication_failures,
            "handover_failures": handover_failures,
            "total_retransmissions": total_retransmissions,
            "nlp_authentication_count": int(
                category_counts.get("Authentication", 0)
            ),
            "nlp_connection_count": int(
                category_counts.get("Connection", 0)
            ),
            "nlp_mobility_count": int(
                category_counts.get("Mobility", 0)
            ),
            "nlp_registration_count": int(
                category_counts.get("Registration", 0)
            ),
            "nlp_retransmission_count": int(
                category_counts.get("Retransmission", 0)
            )
        }

        # Exact training feature order
        model_input = pd.DataFrame(
            [feature_data]
        )[feature_columns]

        prediction = root_cause_model.predict(
            model_input
        )[0]

        result = {
            "ue_id": int(ue_id),
            **feature_data,
            "predicted_root_cause": prediction
        }

        results.append(result)

    return results
'''

inference_path = (
    "/content/drive/MyDrive/"
    "QXDM_AI_Log_Analyzer/src/inference.py"
)

with open(inference_path, "w") as f:
    f.write(inference_code)

print("✅ inference.py created")
print(inference_path)
```
